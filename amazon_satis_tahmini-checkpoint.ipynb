# Amazon Sale Report - Tutar ve Talep Tahmini

Amazon Sale Report - Tutar (Amount) ve Talep (Qty) Tahmini
-----------------------------------------------------------
- Para birimi: veri setinden kontrol edilir (beklenen: INR)
- Amount'un birim fiyat mı toplam tutar mı olduğu veriden doğrulanır
- Leakage'a yol açan sipariş sonrası sütunlar (Status, Courier Status,
  Fulfilment, ship-service-level, fulfilled-by) modele dahil edilmez
- Eğitim/test ayrımı zamana göre yapılır (eski -> eğitim, yeni -> test)
- Amount ve Qty için ayrı modeller, metrik tabloları ve gerçek-tahmin raporları
- Günlük kategori bazlı toplam talep modeli (daha anlamlı talep tahmini)


```python
import warnings

import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

warnings.filterwarnings("ignore")

CSV_PATH = "Amazon Sale Report.csv"
RANDOM_STATE = 42
TEST_RATIO = 0.20
```

```python
# ============================================================
# 1. VERİ YÜKLEME
# ============================================================
df = pd.read_csv(CSV_PATH, low_memory=False)
print("Veri boyutu:", df.shape)
print(df.head())
print("\nSütunlar:", df.columns.tolist())
print("\nEksik değerler:\n", df.isna().sum())
```

```python
# ============================================================
# 2. PARA BİRİMİ KONTROLÜ
# ============================================================
print("\n" + "=" * 60)
print("PARA BİRİMİ KONTROLÜ")
print("=" * 60)
if "currency" in df.columns:
    print(df["currency"].value_counts(dropna=False))
    CURRENCY = df["currency"].mode().iloc[0]
else:
    CURRENCY = "INR"
    print("'currency' sütunu bulunamadı, INR varsayıldı.")
print(f"Raporda kullanılacak para birimi: {CURRENCY}")
```

```python
# ============================================================
# 3. TEMİZLİK
# ============================================================
df = df.dropna(subset=["Amount", "Qty", "Date"]).copy()
df = df[(df["Amount"] > 0) & (df["Qty"] > 0)].copy()  # iptal / boş siparişler

# Tarih formatı bu veri setinde genelde MM-DD-YY (ör. 04-30-22)
parsed = pd.to_datetime(df["Date"], format="%m-%d-%y", errors="coerce")
if parsed.isna().mean() > 0.05:  # format tutmazsa otomatik dene
    parsed = pd.to_datetime(df["Date"], errors="coerce")
df["Date"] = parsed
df = df.dropna(subset=["Date"]).sort_values("Date").reset_index(drop=True)

print("\nTemizlik sonrası boyut:", df.shape)
print("Tarih aralığı:", df["Date"].min().date(), "->", df["Date"].max().date())
```

```python
# ============================================================
# 4. AMOUNT: BİRİM FİYAT MI, TOPLAM TUTAR MI?
# ============================================================
print("\n" + "=" * 60)
print("AMOUNT DOĞRULAMASI (birim fiyat mı, satır toplamı mı?)")
print("=" * 60)

n_multi = int((df["Qty"] > 1).sum())
print(f"Qty > 1 olan satır sayısı: {n_multi} / {len(df)}")

AMOUNT_DEFINITION = "belirsiz (Qty>1 satırı yetersiz)"
if n_multi >= 30 and "SKU" in df.columns:
    chk = df.copy()
    chk["Birim"] = chk["Amount"] / chk["Qty"]
    chk["Grup"] = np.where(chk["Qty"] == 1, "q1", "qN")
    piv = chk.groupby(["SKU", "Grup"]).agg(
        Amount=("Amount", "median"), Birim=("Birim", "median")
    ).unstack("Grup").dropna()
    if len(piv) > 0:
        amount_ratio = (piv[("Amount", "qN")] / piv[("Amount", "q1")]).median()
        unit_ratio = (piv[("Birim", "qN")] / piv[("Birim", "q1")]).median()
        print(f"Karşılaştırılan SKU sayısı          : {len(piv)}")
        print(f"Amount(Qty>1) / Amount(Qty=1) medyan: {amount_ratio:.2f}")
        print(f"Amount/Qty birim fiyat oranı medyan : {unit_ratio:.2f}")
        if abs(unit_ratio - 1) < abs(amount_ratio - 1):
            AMOUNT_DEFINITION = "sipariş satırı toplam tutarı (Qty x birim fiyat)"
        else:
            AMOUNT_DEFINITION = "ürün birim fiyatı"
else:
    print("Qty>1 satırı çok az; kesin sonuç çıkarılamadı.")
    print("Not: Bu veri setinde Qty çoğunlukla 1'dir, bu durumda Amount")
    print("     pratikte satır tutarıdır. Raporda bu sınırlılığı belirt.")

print(f"\n>>> Problem tanımı: Amount = {AMOUNT_DEFINITION} ({CURRENCY})")
```

```python
# ============================================================
# 5. ÖZNİTELİK MÜHENDİSLİĞİ (LEAKAGE'SIZ)
# ============================================================
df["DayOfWeek"] = df["Date"].dt.dayofweek
df["DayOfMonth"] = df["Date"].dt.day
df["IsWeekend"] = (df["DayOfWeek"] >= 5).astype(int)
if "B2B" in df.columns:
    df["B2B"] = df["B2B"].astype(str)

# Çıkarılan (leakage) sütunlar:
#   Status, Courier Status, Fulfilment, ship-service-level, fulfilled-by

candidate_cat = ["Category", "Size", "ship-state", "B2B"]
categorical_features = [c for c in candidate_cat if c in df.columns]
numeric_features = ["DayOfWeek", "DayOfMonth", "IsWeekend"]
features = categorical_features + numeric_features
print("\nKullanılan öznitelikler:", features)
```

```python
# ============================================================
# 6. ZAMAN BAZLI BÖLME 
# ============================================================
cutoff_date = df["Date"].quantile(1 - TEST_RATIO, interpolation="nearest")
train_df = df[df["Date"] <= cutoff_date].copy()
test_df = df[df["Date"] > cutoff_date].copy()

print(f"\nEğitim: {train_df['Date'].min().date()} -> {train_df['Date'].max().date()} ({len(train_df)} satır)")
print(f"Test  : {test_df['Date'].min().date()} -> {test_df['Date'].max().date()} ({len(test_df)} satır)")

X_train, X_test = train_df[features], test_df[features]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            Pipeline([
                ("imp", SimpleImputer(strategy="most_frequent")),
                ("oh", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
            ]),
            categorical_features,
        ),
        (
            "num",
            Pipeline([
                ("imp", SimpleImputer(strategy="median")),
                ("sc", StandardScaler()),
            ]),
            numeric_features,
        ),
    ]
)


def metrics(y_true, y_pred, name):
    return {
        "Model": name,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred),
    }
```

```python
# ============================================================
# 7. AMOUNT TAHMİNİ
# ============================================================
print("\n" + "=" * 60)
print(f"HEDEF: Amount ({CURRENCY})")
print("=" * 60)

y_train_a, y_test_a = train_df["Amount"], test_df["Amount"]

amount_models = {
    "Baseline (Ortalama)": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", DummyRegressor(strategy="mean")),
    ]),
    "Linear Regression": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", TransformedTargetRegressor(
            regressor=LinearRegression(), func=np.log1p, inverse_func=np.expm1)),
    ]),
    "Random Forest": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", RandomForestRegressor(
            n_estimators=200, max_depth=15, min_samples_leaf=5,
            random_state=RANDOM_STATE, n_jobs=-1)),
    ]),
}

res_a, pred_a = [], {}
for name, mdl in amount_models.items():
    mdl.fit(X_train, y_train_a)
    pred_a[name] = mdl.predict(X_test)
    res_a.append(metrics(y_test_a, pred_a[name], name))

amount_table = pd.DataFrame(res_a).set_index("Model").round(3)
print(f"\nAMOUNT METRİK TABLOSU (birim: {CURRENCY})\n", amount_table)
amount_table.to_csv("amount_metrik_tablosu.csv", encoding="utf-8-sig")

# Gerçek vs tahmin raporu (Amount)
amount_report = test_df[["Date"] + categorical_features].copy()
amount_report["Gercek_Amount"] = y_test_a.values
amount_report["Tahmin_LR"] = pred_a["Linear Regression"].round(2)
amount_report["Tahmin_RF"] = pred_a["Random Forest"].round(2)
amount_report["Hata_LR"] = amount_report["Gercek_Amount"] - amount_report["Tahmin_LR"]
amount_report["Hata_RF"] = amount_report["Gercek_Amount"] - amount_report["Tahmin_RF"]
amount_report["MutlakYuzdeHata_RF"] = (
    amount_report["Hata_RF"].abs() / amount_report["Gercek_Amount"] * 100
).round(2)
amount_report.to_csv("gercek_vs_tahmin_amount.csv", index=False, encoding="utf-8-sig")
print("\nGerçek vs Tahmin (ilk 15 satır):")
print(amount_report[["Gercek_Amount", "Tahmin_LR", "Tahmin_RF", "Hata_RF"]].head(15).round(2))

# Kategori bazlı rapor
if "Category" in amount_report.columns:
    cat_report = (
        amount_report.groupby("Category")
        .agg(
            Adet=("Gercek_Amount", "count"),
            Ort_Gercek=("Gercek_Amount", "mean"),
            Ort_Tahmin_RF=("Tahmin_RF", "mean"),
            MAE_RF=("Hata_RF", lambda s: s.abs().mean()),
        )
        .round(2)
        .sort_values("Adet", ascending=False)
    )
    print("\nKategori bazlı rapor:\n", cat_report)
    cat_report.to_csv("kategori_bazli_rapor.csv", encoding="utf-8-sig")

# Grafik: gerçek vs tahmin
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, name in zip(axes, ["Linear Regression", "Random Forest"]):
    p = pred_a[name]
    ax.scatter(y_test_a, p, alpha=0.2, s=8)
    lim = [0, max(y_test_a.max(), p.max())]
    ax.plot(lim, lim, "r--")
    ax.set_title(f"{name} - Amount")
    ax.set_xlabel(f"Gerçek Amount ({CURRENCY})")
    ax.set_ylabel(f"Tahmin ({CURRENCY})")
plt.tight_layout()
plt.savefig("actual_vs_predicted_amount.png", dpi=150)
plt.show()

# Grafik: öznitelik önemi (Random Forest)
rf_a = amount_models["Random Forest"]
names = rf_a.named_steps["pre"].get_feature_names_out()
imps = rf_a.named_steps["m"].feature_importances_
imp = pd.Series(imps, index=names).sort_values(ascending=False).head(15)
plt.figure(figsize=(8, 6))
imp[::-1].plot(kind="barh")
plt.title("Random Forest - En Önemli 15 Öznitelik (Amount)")
plt.xlabel("Önem")
plt.tight_layout()
plt.savefig("feature_importance_amount.png", dpi=150)
plt.show()
```

```python
# ============================================================
# 8. QTY (TALEP) TAHMİNİ - SİPARİŞ SATIRI DÜZEYİ
# ============================================================
print("\n" + "=" * 60)
print("HEDEF: Qty (Talep Adedi) - sipariş satırı düzeyi")
print("=" * 60)
print("Qty dağılımı:\n", df["Qty"].value_counts(normalize=True).head().round(4))

y_train_q, y_test_q = train_df["Qty"], test_df["Qty"]

qty_models = {
    "Baseline (Ortalama)": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", DummyRegressor(strategy="mean")),
    ]),
    "Baseline (Medyan)": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", DummyRegressor(strategy="median")),
    ]),
    "Linear Regression": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", LinearRegression()),
    ]),
    "Random Forest": Pipeline([
        ("pre", clone(preprocessor)),
        ("m", RandomForestRegressor(
            n_estimators=200, max_depth=10, min_samples_leaf=10,
            random_state=RANDOM_STATE, n_jobs=-1)),
    ]),
}

res_q, pred_q = [], {}
for name, mdl in qty_models.items():
    mdl.fit(X_train, y_train_q)
    pred_q[name] = mdl.predict(X_test)
    r = metrics(y_test_q, pred_q[name], name)
    r["Dogruluk(yuvarlanmis)"] = float((np.round(pred_q[name]) == y_test_q).mean())
    res_q.append(r)

qty_table = pd.DataFrame(res_q).set_index("Model").round(4)
print("\nQTY METRİK TABLOSU\n", qty_table)
qty_table.to_csv("qty_metrik_tablosu.csv", encoding="utf-8-sig")

qty_report = test_df[["Date"] + categorical_features].copy()
qty_report["Gercek_Qty"] = y_test_q.values
qty_report["Tahmin_RF"] = pred_q["Random Forest"].round(3)
qty_report["Hata"] = qty_report["Gercek_Qty"] - qty_report["Tahmin_RF"]
qty_report.to_csv("gercek_vs_tahmin_qty.csv", index=False, encoding="utf-8-sig")

plt.figure(figsize=(6, 5))
plt.scatter(y_test_q, pred_q["Random Forest"], alpha=0.2, s=8)
lim = [0, max(y_test_q.max(), pred_q["Random Forest"].max())]
plt.plot(lim, lim, "r--")
plt.title("Random Forest - Qty: Gerçek vs Tahmin")
plt.xlabel("Gerçek Qty")
plt.ylabel("Tahmin Qty")
plt.tight_layout()
plt.savefig("actual_vs_predicted_qty.png", dpi=150)
plt.show()
```

```python
# ============================================================
# 9. GÜNLÜK KATEGORİ BAZLI TOPLAM TALEP TAHMİNİ
# ============================================================
print("\n" + "=" * 60)
print("HEDEF: Günlük kategori bazlı toplam talep (ToplamTalep)")
print("=" * 60)

daily = (
    df.groupby(["Date", "Category"], as_index=False)["Qty"].sum()
    .rename(columns={"Qty": "ToplamTalep"})
)
daily["DayOfWeek"] = daily["Date"].dt.dayofweek
daily["DayOfMonth"] = daily["Date"].dt.day
daily["IsWeekend"] = (daily["DayOfWeek"] >= 5).astype(int)

d_train = daily[daily["Date"] <= cutoff_date]
d_test = daily[daily["Date"] > cutoff_date]
d_feats = ["Category", "DayOfWeek", "DayOfMonth", "IsWeekend"]

d_pre = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), ["Category"]),
    ("num", "passthrough", ["DayOfWeek", "DayOfMonth", "IsWeekend"]),
])

daily_models = {
    "Baseline (Ortalama)": Pipeline([
        ("pre", clone(d_pre)), ("m", DummyRegressor(strategy="mean"))]),
    "Linear Regression": Pipeline([
        ("pre", clone(d_pre)), ("m", LinearRegression())]),
    "Random Forest": Pipeline([
        ("pre", clone(d_pre)),
        ("m", RandomForestRegressor(
            n_estimators=200, min_samples_leaf=3,
            random_state=RANDOM_STATE, n_jobs=-1))]),
}

res_d, pred_d = [], {}
for name, mdl in daily_models.items():
    mdl.fit(d_train[d_feats], d_train["ToplamTalep"])
    pred_d[name] = mdl.predict(d_test[d_feats])
    res_d.append(metrics(d_test["ToplamTalep"], pred_d[name], name))

daily_table = pd.DataFrame(res_d).set_index("Model").round(3)
print("\nGÜNLÜK TALEP METRİK TABLOSU\n", daily_table)
daily_table.to_csv("gunluk_talep_metrik_tablosu.csv", encoding="utf-8-sig")

daily_report = d_test[["Date", "Category", "ToplamTalep"]].copy()
daily_report["Tahmin_RF"] = pred_d["Random Forest"].round(2)
daily_report["Hata"] = daily_report["ToplamTalep"] - daily_report["Tahmin_RF"]
daily_report.to_csv("gercek_vs_tahmin_gunluk_talep.csv", index=False, encoding="utf-8-sig")

# Grafik: toplam günlük talep, gerçek vs tahmin
tot = (
    daily_report.groupby("Date")[["ToplamTalep", "Tahmin_RF"]].sum().sort_index()
)
plt.figure(figsize=(11, 4.5))
plt.plot(tot.index, tot["ToplamTalep"], label="Gerçek", marker="o", ms=3)
plt.plot(tot.index, tot["Tahmin_RF"], label="Tahmin (RF)", marker="o", ms=3)
plt.title("Test Dönemi - Günlük Toplam Talep: Gerçek vs Tahmin")
plt.xlabel("Tarih")
plt.ylabel("Toplam adet")
plt.legend()
plt.tight_layout()
plt.savefig("gunluk_talep_gercek_vs_tahmin.png", dpi=150)
plt.show()
```
